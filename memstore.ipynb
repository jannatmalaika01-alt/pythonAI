```python
from langgraph.store.memory import InMemoryStore
```

```python
#creqating a memory store
store=InMemoryStore()
```

```python
#creating a namespace 
namespace=("user","u1")
```

```python
#Creating memories
store.put(namespace,"1",{"data":"user likes pizza"})
store.put(namespace,"2",{"data":"user likes dark mode"})
```

```python
#creating another namespace
namespace2=("user","u2")
store.put(namespace2,"1",{"data":"user likes icecream"})
store.put(namespace2,"2",{"data":"user likes hamburger menu"})
```

```python
#retrieve memories
store.get(namespace,"2")
```

```python
print(store.get(namespace, "1"))
print(store.get(namespace, "2"))

print(store.get(namespace2, "1"))
print(store.get(namespace2, "2"))
```

```python
#searching all memories for user 1
items=store.search(namespace)
for item in items:
    print(item.value)
```

SEMANTIC SEARCHING

```python
from langchain_google_genai import GoogleGenerativeAIEmbeddings
from dotenv import load_dotenv
```

```python
load_dotenv
```

```python
embeddings = GoogleGenerativeAIEmbeddings(
    model="models/gemini-embedding-001"
)
```

```python
store=InMemoryStore(index={"embed":embeddings,"dims":1536})
```

```python
namespace=("user","u1")
```

```python
store.put(namespace,"1",{"data":"user likes pastry"})
store.put(namespace,"2",{"data":"user prefers dark mode"})
store.put(namespace,"3",{"data":"user likes to travel"})
store.put(namespace,"4",{"data":"user wants scholarship"})
store.put(namespace,"5",{"data":"user is learning french"})
store.put(namespace,"6",{"data":"user is learning langchain,agentic ai,python"})
```

```python
items=store.search(namespace,query="what is user currently learning?",limit=2)
for item in items:
    print(item.value)
```
